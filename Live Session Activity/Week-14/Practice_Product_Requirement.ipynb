```python
import warnings
warnings.filterwarnings('ignore')
import langchain, langchain_community
print(langchain.__version__)
print(langchain_community.__version__)

import os
import json
from langchain_openai import ChatOpenAI
from langchain.schema import HumanMessage, SystemMessage
from textwrap import indent

import utils
```

```python
llm=ChatOpenAI(model="gpt-3.5-turbo", temperature=0)
              
```

```python
#Here is a prompt designed for a hypothetical product: a Smart AI-Powered Calendar App.
  
prompt = f"""
    Act as a Principal Product Manager. Write a comprehensive Product Requirements Document (PRD) 
    for a Smart AI-Powered Calendar App that automatically schedules meetings based on user energy levels and habits. 
    
    CRITICAL: The entire response must be strictly under 200 words. Keep it highly concise.

    Structure your response using these exact sections:
    1. Executive Summary: Core value proposition.
    2. Problem Statement: Main user pain points.
    3. User Stories & Features: One high-priority user story in 'As a... I want to... So that...' format with brief acceptance criteria.
    4. Technical Constraints: Data privacy (GDPR) and API integration limits.
    5. Success Metrics: Two measurable Key Performance Indicators (KPIs).
    """
    


```

```python
dynamic_messages = [
    SystemMessage(content="You are a helpful assistant."),
    HumanMessage(content=prompt)
]
dynamic_response = llm(dynamic_messages).content.strip()
```

```python
print("Dynamic PROMPT:\n",indent(dynamic_response, "    "))
```
