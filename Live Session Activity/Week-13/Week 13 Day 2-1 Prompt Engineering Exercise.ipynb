```python
import os
import json
from langchain_openai import ChatOpenAI
from langchain.schema import HumanMessage, SystemMessage
from textwrap import indent
import openai

import utils
```

```python
llm = ChatOpenAI(model="gpt-3.5-turbo", temperature=0)
```

```python
def get_response(prompt, model=llm):
    messages = [SystemMessage(content="You are a helpful assistant"),
                HumanMessage(content=prompt)
               ]
    response = model.invoke(messages).content.strip()
    return response
```

**Zero Shot Prompting**

```python
prompt = """
Translate the following English sentence to French:

'Hello, world!'
"""
get_response(prompt)
```

```python
prompt = """
Classify the following statements as Positive or Negative sentiment.

'Great restaurant! Good food, good service. Would love to recommend.'
"""

get_response(prompt)
```

**One Shot Prompting**

```python
prompt = """
You are required to predict the sentiment of customer feedback.

Feedback: "I loved the quick service and friendly staff."
Classification: Positive

Feedback: "We had a great experience. I really enjoyed the food."
Classification:
"""
```

```python
get_response(prompt)
```

**Few Shot Prompting**

```python
prompt = """
You are required to predict the sentiment of customer feedback.

Feedback: "I loved the quick service and friendly staff."
Classification: Positive

Feedback: "The product did not meet my expectations."
Classification: Negative

Feedback: "I am not sure if this is the right product for me."
Classification: Neutral

Feedback: "We had a great experience. I really enjoyed the food."
Classification:
"""
```

```python
get_response(prompt)
```

**Chain of Thought**

```python
xprompt = """ Question: Roger has 5 tennis balls. He buys 2 more cans of tennis balls. Each can has 3 tennis balls.
How many tennis balls does he have now?

Answer: Roger started with 5 balls. 2 cans of 3 tennis balls each is 6 tennis balls. 5 + 6 = 11. Hence he has 11 tennis balls.

Question: The cafeteria has 23 apples. They used 20 apples to make lunch. They bought 6 more apples. How many apples are still left?
"""
```

```python
print(get_response(xprompt))
```
