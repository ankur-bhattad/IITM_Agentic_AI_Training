```python
# ============================================================
#  Properties of Sets
# ============================================================

# ----------------------
# Unordered (no indexing)
# ----------------------
fruits = {'apple', 'banana', 'cherry'}

# This will raise an error because sets don't support indexing:
# print(fruits[0])   # ❌ TypeError


# ----------------------
# Unique Elements
# ----------------------
numbers = set([1, 2, 2, 3, 3, 3])
print(numbers)       # Output: {1, 2, 3}


# ----------------------
# Mutable but with Restrictions
# ----------------------

# Works because tuples are immutable
s = {1, 2, (3, 4)}

# Fails because lists are mutable
# s = {1, 2, [3, 4]}   # ❌ TypeError: unhashable type: 'list'
```

```python
# ============================================================
#  Common Operations on Sets
# ============================================================

# Starting set
fruits = {'apple', 'banana', 'cherry'}

# Adding elements
fruits.add('orange')        # Result: {'apple', 'banana', 'cherry', 'orange'}

# Removing elements (error if not found)
# fruits.remove('banana')   # Works
# fruits.remove('pear')     # ❌ KeyError

# Remove without error
fruits.discard('pear')      # No error if not found

# Checking membership
print('apple' in fruits)    # Output: True

# Length of a set
print(len(fruits))          # Example output: 3


# Other methods:
# pop()      → removes and returns an arbitrary element
# clear()    → removes all elements
# copy()     → returns a shallow copy
```

```python
# ============================================================
#  Mathematical Operations on Sets
# ============================================================

set1 = {1, 2, 3}
set2 = {3, 4, 5}

# ----------------------
# Union
# ----------------------
print(set1 | set2)               # Output: {1, 2, 3, 4, 5}
print(set1.union(set2))          # Same result


# ----------------------
# Intersection
# ----------------------
print(set1 & set2)               # Output: {3}
print(set1.intersection(set2))   # Same result


# ----------------------
# Difference
# ----------------------
print(set1 - set2)               # Output: {1, 2}
print(set1.difference(set2))     # Same result


# ----------------------
# Symmetric Difference
# ----------------------
print(set1 ^ set2)                      # Output: {1, 2, 4, 5}
print(set1.symmetric_difference(set2))  # Same result

```

```python
# ============================================================
#  Use Cases of Sets
# ============================================================

# ----------------------
# Removing Duplicates
# ----------------------
items = ['apple', 'banana', 'apple', 'orange']
unique_items = set(items)        # Output: {'orange', 'banana', 'apple'}


# ----------------------
# Membership Testing
# ----------------------
fruits = {'apple', 'banana', 'cherry'}
print('banana' in fruits)        # Output: True


# ----------------------
# Finding Unique or Common Elements
# ----------------------
visitors_day1 = {'Alice', 'Bob', 'Charlie'}
visitors_day2 = {'Bob', 'David', 'Eve'}

# Common visitors (intersection)
returning = visitors_day1 & visitors_day2   # Output: {'Bob'}

# All visitors (union)
all_visitors = visitors_day1 | visitors_day2
```

```python
#  The if-else Statement

x = 3

if x % 2 == 0:
    print("Even number")
else:
    print("Odd number")

# Explanation:
# - The condition (x % 2 == 0) is checked.
# - If True → run the if block.
# - If False → run the else block.
# - if-else ensures one of the two blocks always executes.
```

```python
#  The if-elif-else Statement

score = 85

if score >= 90:
    print("Grade A")
elif score >= 75:
    print("Grade B")
elif score >= 60:
    print("Grade C")
else:
    print("Grade D")

# Explanation:
# - Conditions are checked in order, top to bottom.
# - The first condition that evaluates to True executes.
# - Once a condition matches, the remaining conditions are skipped.
# - The else block runs only if all previous conditions are False.
```

```python
#  Conditional Expressions (Ternary Operator)

# Standard if-else
x = 7
if x % 2 == 0:
    result = "Even"
else:
    result = "Odd"
print(result)

# Ternary operator version
result = "Even" if x % 2 == 0 else "Odd"
print(result)
```

```python
# ---------------------------------------
#  Scope of Variables in Functions
# ---------------------------------------

def my_function():
    local_var = 10          # Local scope
    print("Inside function, local_var =", local_var)

global_var = 20            # Global scope

my_function()

print("Outside function, global_var =", global_var)   # Output: 20

# The next line will cause an error because local_var is not accessible outside the function
# print(local_var)   # NameError: name 'local_var' is not defined
```

```python
# -----------------------------
#  Lambda Functions
# -----------------------------

# A lambda function that returns square of a number
square = lambda x: x ** 2

print(square(4))   # Output: 16
```

```python
# Simple chatbot using functions

def get_response(user_input):
    """Return a chatbot response based on user input"""
    user_input = user_input.lower()

    if "hello" in user_input or "hi" in user_input:
        return "Hello! How can I help you today?"
    elif "bye" in user_input:
        return "Goodbye! Have a great day!"
    elif "price" in user_input:
        return "Our product prices start at ₹499."
    else:
        return "I'm not sure I understand. Can you rephrase?"

# Chatbot loop
print("Welcome to SimpleBot! (type 'bye' to exit)")

while True:
    user_message = input("You: ")
    response = get_response(user_message)
    print("Bot:", response)

    if "bye" in user_message.lower():
        break
```
