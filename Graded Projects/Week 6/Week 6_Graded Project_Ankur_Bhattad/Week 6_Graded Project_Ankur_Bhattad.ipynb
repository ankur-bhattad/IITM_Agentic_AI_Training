2. Tokenization Demo

```python
from transformers import AutoTokenizer

# Sentence to tokenize
sentence = "LLMs are powerful tools for natural language understanding."

# Load a free, small tokenizer (BERT)
tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")

# Tokenize the sentence
tokens = tokenizer.tokenize(sentence)

# Convert tokens to token IDs
token_ids = tokenizer.convert_tokens_to_ids(tokens)

# Get encoded input (includes special tokens like [CLS], [SEP])
encoded = tokenizer(sentence)

# Print outputs
print("Original Sentence:")
print(sentence)

print("\nTokens:")
print(tokens)

print("\nToken IDs:")
print(token_ids)

print("\nSequence Length (including special tokens):")
print(len(encoded["input_ids"]))
```

Section B: Prompt Engineering

3. Prompt Tuning Challenge

```python
from transformers import pipeline

# Load a small free model
generator = pipeline("text2text-generation",  model="google/flan-t5-large")

# Prompts
prompts = [
    "summarize: Artificial Intelligence is transforming industries by automating tasks, improving decision-making, and enabling new innovations across sectors.",
    "question: What is the capital of France?",
    "Write a poem about AI."
]

# Generate outputs
for p in prompts:
    result = generator(p, max_length=60, do_sample=True, temperature=0.7, top_p=0.9)
    
    print("\nPrompt:", p)
    print("Output:", result[0]['generated_text'])
```

```python
from transformers import pipeline

# Load a small free model
generator = pipeline("text2text-generation",  model="google/flan-t5-large")

# Prompts
prompts = [
    "extract keywords: Artificial Intelligence is transforming industries by automating tasks, improving decision-making, and enabling new innovations across sectors.",
    "question: What is the capital of France? answer in 3 words.",
    "Write a 4-line poem about AI where each line starts with AI."
]

# Generate outputs
for p in prompts:
    result = generator(p, max_length=60, do_sample=True, temperature=0.7, top_p=0.9)
    print("\nPrompt:", p)
    print("Output:", result[0]['generated_text'])
```

Reflection on Prompt Tuning:

Rephrasing prompts improved output quality and consistency. More specific instructions led to better adherence to constraints such as word limits and formatting. 
For summarisation, explicitly mentioning the word limit and focus helped reduce unnecessary details. 
In Q&A tasks, instructing the model to provide brief, factual responses resulted in more precise answers compared to generic prompts. 
For creative generation, adding structural guidance ensured the output followed the desired format, whereas vague prompts produced inconsistent results. 
Overall, prompt clarity, constraints, and formatting instructions play a crucial role in guiding LLM behaviour. 

However, lot depends on choosing appropriate model as well. Even small changes in wording can shift outputs from generic and verbose to precise and task-aligned in large models but when it comes to small models, it seems sometimes these models are ignoring slight variation in prompt.

```python
import gensim.downloader as api
# Load pre-trained GloVe embeddings (50-dimensional)
model = api.load("glove-wiki-gigaword-50")
words = ["king", "queen", "diamond"]
for word in words:
    print(f"\nWord: {word}")
    
    # a. First 10 values of vector
    vector = model[word]
    print("First 10 vector values:")
    print(vector[:10])
    
    # b. Top 5 most similar words
    print("\nTop 5 similar words:")
    similar_words = model.most_similar(word, topn=5)
    
    for sim_word, score in similar_words:
        print(f"{sim_word}: {score:.4f}")
```

Section C: Embeddings with Gensim

5. Sentence Embeddings

```python
!pip install gensim scikit-learn
```

```python
import gensim.downloader as api
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity
model = api.load("glove-wiki-gigaword-50")
sentences = [
    "Artificial intelligence is transforming the world",
    "Machine learning is a subset of AI",
    "Deep learning uses neural networks",
    "Gold and diamond jewellery are valuable",
    "Jewellery is often made from precious metals"
]
```

```python
def sentence_embedding(sentence, model):
    words = sentence.lower().split()
    vectors = []
    
    for word in words:
        if word in model:
            vectors.append(model[word])
    
    if len(vectors) == 0:
        return np.zeros(model.vector_size)
    
    return np.mean(vectors, axis=0)
```

```python
embeddings = []

for sentence in sentences:
    emb = sentence_embedding(sentence, model)
    embeddings.append(emb)

embeddings = np.array(embeddings)

similarity_matrix = cosine_similarity(embeddings)

print("Cosine Similarity Matrix:\n")
print(similarity_matrix)
```

Transformer Application

Sentiment Classification using Hugging Face pipeline

```python
from transformers import pipeline

# Load sentiment analysis pipeline
classifier = pipeline("sentiment-analysis")

# Custom inputs
texts = [
    "The product quality is excellent and delivery was super fast!",
    "I am very disappointed with the customer service.",
    "The experience was okay, nothing special but not bad either.",
    "Absolutely loved the user interface and ease of use!",
    "The app keeps crashing and it's very frustrating."
]

# Run predictions
results = classifier(texts)

# Display results
for text, result in zip(texts, results):
    print(f"Text: {text}")
    print(f"Sentiment: {result['label']}")
    print("-" * 50)
```

Reflection:

Sentiment classification using Transformers can significantly improve how businesses handle customer feedback at scale. 
Instead of manually reviewing thousands of reviews, companies can automatically categorize feedback into positive, negative, or neutral sentiments in real time. 
This enables faster response to dissatisfied customers, helping improve retention and brand reputation. 
It can also identify common pain points, such as product issues or service delays, by analyzing trends in negative sentiment. 
Additionally, marketing teams can use positive feedback to highlight strengths in campaigns. Overall, this application enhances decision-making, operational efficiency, and customer experience across professional workflows.
