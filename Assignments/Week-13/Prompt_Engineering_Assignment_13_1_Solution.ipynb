# Practice Assignment 13.1 – Prompt Engineering

This notebook demonstrates all required prompting techniques:
- Role-based
- Instruction-based
- Chain-of-Thought (requesting reasoning while instructing the model to provide concise final reasoning)
- Few-shot
- Constraint-based
- Prompt refinement

Each scenario includes:
1. Baseline Prompt (V1)
2. Sample Output
3. Analysis
4. Refined Prompt (V2)
5. Improved Output
6. Reflection

> Note: Sample outputs are representative examples for learning purposes.

```python
import os
import json
from langchain_openai import ChatOpenAI
from langchain.schema import HumanMessage, SystemMessage
from textwrap import indent
import openai
from IPython.display import Markdown, display

import utils
```

```python
llm = ChatOpenAI(model="gpt-3.5-turbo", temperature=0)
```

```python
def get_response(prompt, model=llm):
    messages = [SystemMessage(content="You are a Management Consultant"),
                HumanMessage(content=prompt)
               ]
    response = model.invoke(messages).content.strip()
    return response
```

# Scenario 1: Business Strategy


```python
## Prompt Version 1

prompt = """
Develop a 12-month omnichannel strategy for a retail company with declining store footfall but growing online sales.
"""

display(Markdown(get_response(prompt)))
```

```python
## Prompt Version 2
prompt = """
Act as a Management Consultant.

Task:
Develop a 12-month omnichannel strategy for a retail company with declining store footfall but growing online sales.

Constraints:
- Existing workforce only
- Moderate budget
- Timeline: 12 months
- Mention risks and KPIs

Think through the problem carefully, then provide only the final structured answer.

Output format:
1. Situation Analysis
2. Root Causes
3. Recommendations
4. Roadmap Table
5. Risks
6. KPIs
"""
display(Markdown(get_response(prompt)))
```
