# Practice Assignment 11.1: Hierarchical Planning with CrewAI

## Corporate Event Planning Automation

This notebook implements a **Hierarchical Planning AI System** using CrewAI for corporate event planning. The workflow follows:

1. Event Planner Agent (Hierarchical Planning)
2. Vendor Research Agent
3. Budget Analyst Agent
4. Schedule Designer Agent
5. Final Event Coordinator Agent

The system accepts dynamic inputs and generates:
- Hierarchical Event Plan
- Vendor Research Summary
- Budget Estimate
- Full-Day Event Schedule
- Consolidated Final Event Plan

```
import os
from crewai import Agent, Task, Crew, Process
from langchain_openai import ChatOpenAI

# Set your API key before running
# os.environ["OPENAI_API_KEY"] = "YOUR_API_KEY"

llm = ChatOpenAI(
    model="gpt-4o-mini",
    temperature=0.2
)
```

## Step 1 - Define Agents

```
event_planner = Agent(
    role="Event Planner Agent",
    goal="Create a hierarchical event planning strategy for a corporate event.",
    backstory="""You are a senior event strategist who breaks complex event goals
    into structured plans, milestones, dependencies and success metrics.""",
    llm=llm,
    verbose=True
)

vendor_researcher = Agent(
    role="Vendor Research Agent",
    goal="Research and recommend suitable vendors for the event.",
    backstory="""You specialize in identifying vendors for catering,
    audio visual support, decor, seating and stage setup.""",
    llm=llm,
    verbose=True
)

budget_analyst = Agent(
    role="Budget Analyst Agent",
    goal="Build detailed event budgets and identify optimization opportunities.",
    backstory="""You are an event finance expert skilled at estimating
    costs and prioritizing critical spending categories.""",
    llm=llm,
    verbose=True
)

schedule_designer = Agent(
    role="Schedule Designer Agent",
    goal="Create a realistic full-day event schedule.",
    backstory="""You design conference agendas balancing keynotes,
    workshops, networking sessions and operational buffers.""",
    llm=llm,
    verbose=True
)

final_coordinator = Agent(
    role="Final Event Coordinator Agent",
    goal="Consolidate all outputs into a final event execution plan.",
    backstory="""You are an experienced event director responsible for
    turning planning artifacts into a production-ready event plan.""",
    llm=llm,
    verbose=True
)
```

## Step 2 - Define Tasks

```
planning_task = Task(
    description="""
    Event Goal: {event_goal}
    Date: {date}
    Attendees: {attendee_count}
    Preferences: {preferences}

    Create a hierarchical event plan:
    Level 1 -> Strategic pillars
    Level 2 -> Major workstreams
    Level 3 -> Detailed execution tasks

    Include dependencies, milestones and success metrics.
    """,
    expected_output="Hierarchical event plan with Level 1, Level 2, Level 3 breakdown and success metrics.",
    agent=event_planner
)

vendor_task = Task(
    description="""
    Using the event context:
    Goal: {event_goal}
    Attendees: {attendee_count}
    Preferences: {preferences}

    Research and recommend:
    - Catering vendors (2-3 options)
    - Audio/Visual vendors
    - Seating and decor providers
    - Optional stage setup vendors

    Provide strengths, limitations and suitability.
    """,
    expected_output="Vendor research report with categorized recommendations.",
    agent=vendor_researcher
)

budget_task = Task(
    description="""
    Based on the event goal, attendee count and vendor recommendations,
    create a detailed budget.

    Categories:
    - Venue
    - Food & Beverage
    - Decor
    - Audio/Visual
    - Labor
    - Marketing
    - Contingency

    Include total estimate and cost-saving opportunities.
    """,
    expected_output="Budget table with category estimates, total cost and cost optimization suggestions.",
    agent=budget_analyst
)

schedule_task = Task(
    description="""
    Create a realistic full-day schedule for:

    Goal: {event_goal}
    Date: {date}
    Attendees: {attendee_count}

    Include:
    - Registration
    - Keynotes
    - Workshops
    - Breaks
    - Networking sessions
    - Speaker preparation windows
    - Buffer times
    """,
    expected_output="Detailed event-day agenda with time slots.",
    agent=schedule_designer
)

final_plan_task = Task(
    description="""
    Consolidate all prior outputs into one final event plan.

    Include:
    - Executive Summary
    - Hierarchical Plan
    - Vendor Summary
    - Budget Summary
    - Event Schedule
    - Booking Checklist
    - Risks and Contingencies
    - Final Recommendations
    """,
    expected_output="Comprehensive final event plan document.",
    agent=final_coordinator
)
```

## Step 3 - Build the Crew

```
event_crew = Crew(
    agents=[
        event_planner,
        vendor_researcher,
        budget_analyst,
        schedule_designer,
        final_coordinator
    ],
    tasks=[
        planning_task,
        vendor_task,
        budget_task,
        schedule_task,
        final_plan_task
    ],
    process=Process.sequential,
    verbose=True
)
```

## Step 4 - Execute with Dynamic Inputs

```
sample_inputs = {
    "event_goal": "Organize a 1-day corporate innovation summit focusing on AI and automation.",
    "date": "2025-11-15",
    "attendee_count": "300 attendees",
    "preferences": "Premium decor, vegetarian+vegan meal options, mid-range budget, interactive sessions"
}

result = event_crew.kickoff(inputs=sample_inputs)

print(result)
```
